```python
import numpy as np
wavelength_m = 1550e-9
transmit_aperture_diameter = 0.3
distance2 = 20.0
transmit_power2 = 1.0
aperture_diameter2 = 2.0
aperture_area2 = np.pi * (aperture_diameter2 / 2)**2
theta = wavelength_m / transmit_aperture_diameter
spot_radius = distance2 * theta
spot_area = np.pi * spot_radius**2

if aperture_area2 > spot_area:
    received_power_laser = transmit_power2
else:
    received_power_laser = transmit_power2 * (aperture_area2 / spot_area)

print(received_power_laser)
```
