```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from src.channel.poisson_channel import sample_photon_counts

lambda_test = 0.55
samples = sample_photon_counts(lambda_test, size=10000, seed=42)

print(f"Sample mean: {np.mean(samples):.4f}")
print(f"Sample variance: {np.var(samples):.4f}")
```

```python
values, counts = np.unique(samples, return_counts=True)
observed_freq = counts / len(samples)

k_range = np.arange(0, values.max() + 1)
theoretical_pmf = poisson.pmf(k_range, lambda_test)

plt.figure(figsize=(8, 5))
plt.bar(values, observed_freq, alpha=0.6, label="Observed (simulation)")
plt.plot(k_range, theoretical_pmf, 'ro-', label="Theoretical Poisson PMF")
plt.xlabel("Photon count (k)")
plt.ylabel("Probability")
plt.title(f"Poisson distribution validation (λ={lambda_test})")
plt.legend()
plt.savefig("../figures/poisson_validation_pmf.png")
plt.show()
```
